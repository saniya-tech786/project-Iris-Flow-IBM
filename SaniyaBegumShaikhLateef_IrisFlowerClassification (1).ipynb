## 1. Project Overview
This project explores the classic Iris flower dataset and trains a machine-learning model to classify an iris flower into one of three species: *setosa*, *versicolor*, or *virginica*.

## 2. Objectives
- Load and inspect the dataset.
- Explore the features and species distribution.
- Split data into training and testing sets.
- Train a classification model.
- Evaluate the model using accuracy, a classification report, and a confusion matrix.
- Demonstrate a prediction for a new flower measurement.

## 3. Dataset
The Iris dataset contains 150 observations and four numeric features:
- Sepal length (cm)
- Sepal width (cm)
- Petal length (cm)
- Petal width (cm)

Dataset source: [UCI Machine Learning Repository — Iris](https://archive.ics.uci.edu/dataset/53/iris)

For convenience, this notebook loads the copy bundled with scikit-learn, so a separate dataset download is not required.


```python
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

RANDOM_STATE = 42
```

## 4. Load the Dataset

```python
iris = load_iris(as_frame=True)
df = iris.frame.copy()
df["species"] = df["target"].map(dict(enumerate(iris.target_names)))

df.head()
```

## 5. Understand the Data

```python
print("Dataset shape:", df.shape)
print("\\nColumn information:")
df.info()
print("\\nMissing values per column:")
print(df.isnull().sum())
print("\\nSummary statistics:")
display(df.describe())
```

## 6. Exploratory Data Analysis

```python
# Species distribution
plt.figure(figsize=(7, 4))
sns.countplot(data=df, x="species")
plt.title("Number of Samples per Iris Species")
plt.xlabel("Species")
plt.ylabel("Count")
plt.tight_layout()
plt.show()

# Pairwise relationships among measurements
sns.pairplot(df, vars=iris.feature_names, hue="species", diag_kind="hist")
plt.show()
```

## 7. Prepare Data and Train the Model

```python
feature_columns = iris.feature_names
X = df[feature_columns]
y = df["target"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=RANDOM_STATE, stratify=y
)

# Scaling is included in a pipeline to keep preprocessing fitted only on training data.
model = make_pipeline(
    StandardScaler(),
    KNeighborsClassifier(n_neighbors=5)
)
model.fit(X_train, y_train)

print(f"Training samples: {len(X_train)}")
print(f"Testing samples: {len(X_test)}")
```

## 8. Evaluate the Model

```python
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)

print(f"Test accuracy: {accuracy:.2%}")
print("\\nClassification report:")
print(classification_report(y_test, y_pred, target_names=iris.target_names))

cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(6, 4))
sns.heatmap(
    cm, annot=True, fmt="d", cmap="Blues",
    xticklabels=iris.target_names,
    yticklabels=iris.target_names
)
plt.title("Confusion Matrix")
plt.xlabel("Predicted Species")
plt.ylabel("Actual Species")
plt.tight_layout()
plt.show()
```

## 9. Predict a New Flower

```python
# Example measurements in centimeters:
# sepal length, sepal width, petal length, petal width
new_flower = pd.DataFrame(
    [[5.1, 3.5, 1.4, 0.2]],
    columns=feature_columns
)

predicted_class = model.predict(new_flower)[0]
print("Predicted species:", iris.target_names[predicted_class])
```
